```python
from PIL import Image
from IPython import display
```

```python
import cv2
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('anapa.jpg')
Image.fromarray(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
```

```python
image_grayscale = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
Image.fromarray(image_grayscale)
```

```python
red_hist = [0] * 256
green_hist = [0] * 256
blue_hist = [0] * 256
total_pixels = 0
for row in image:
    for color in row:
        red_hist[color[0]] += 1
        green_hist[color[1]] += 1
        blue_hist[color[2]] += 1
        total_pixels += 1

plt.bar(range(256), red_hist, color=(1.0, 0, 0, 1.0))
plt.bar(range(256), green_hist, color=(0.0, 1.0, 0, 1.0))
plt.bar(range(256), blue_hist, color=(0.0, 0, 1.0, 1.0))

max_brightness_count = 0
zero_brightness_count = 0
common_brightness_count = 0
for row in image_grayscale:
    for color in row:
        if color == 0:
            zero_brightness_count += 1
        elif color == 255:
            max_brightness_count += 1
        elif color > 255 * 0.05 and color < 255 * 0.95:
            common_brightness_count += 1

print(f"Total pixels: {total_pixels}")
print(f"Max brightness: {max_brightness_count} ({max_brightness_count / total_pixels:.2f}%)")
print(f"Zero brightness: {zero_brightness_count} ({zero_brightness_count / total_pixels:.2f}%)")
print(f">5% and <95% brightness: {common_brightness_count} ({common_brightness_count / total_pixels:.2f}%)")

plt.show()
```
